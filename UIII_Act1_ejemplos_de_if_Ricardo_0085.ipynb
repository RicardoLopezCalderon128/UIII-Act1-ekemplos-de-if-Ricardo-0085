#### ejemplo de if

```python
# 1. El condicional simple if
# Se utiliza cuando solo queremos realizar una acción si se cumple una condición específica
#.
# Ejemplo: Verificar si un número es par Este ejemplo utiliza el operador aritmético de módulo (%) para obtener el residuo y el operador relacional de igualdad (==)
#.
numero = 10 # 1. Se asigna un valor entero a la variable.
if numero % 2 == 0: # 2. Si el residuo de numero/2 es igual a 0.
    print("El número es par.") # 3. Se imprime si la condición es verdadera.
```

```python
# 2. El condicional if-else
# Se utiliza para elegir entre dos alternativas posibles: una si la condición es verdadera y otra si es falsa
# .
# Ejemplo: Determinar si una persona es mayor de edad Utiliza el operador relacional mayor o igual que (>=)
# .
edad = int(input("Ingresa tu edad: ")) # 1. Pide la edad y la convierte a entero.
if edad >= 18: # 2. Compara si la edad es mayor o igual a 18.
    print("Eres mayor de edad.") # 3. Acción si es verdadero.
else: # 4. Si la condición anterior fue falsa.
    print("Eres menor de edad.") # 5. Acción si es falso.
```

```python
# 3. El condicional if-elif-else
# Permite evaluar múltiples condiciones en orden. En cuanto una se cumple, las demás se omiten
# .
# Ejemplo: Clasificación de un número (Positivo, Negativo o Cero) Utiliza los operadores relacionales mayor que (>) y menor que (<)
# .
numero = int(input("ingresa un numero: ")) # 1. Asignación de un número negativo.
if numero > 0: # 2. Primera condición: ¿es mayor a cero?
    print("El número es positivo.") # 3. Ejecuta si es mayor a 0.
elif numero < 0: # 4. Segunda condición: ¿es menor a cero?
    print("El número es negativo.") # 5. Ejecuta si es menor a 0.
else: # 6. Si no es mayor ni menor a cero.
    print("El número es cero.") # 7. Ejecuta si es exactamente 0.
```

```python
# 3. El condicional if-elif-else
# Permite evaluar múltiples condiciones en orden. En cuanto una se cumple, las demás se omiten
# .
# Ejemplo: Clasificación de un número (Positivo, Negativo o Cero) Utiliza los operadores relacionales mayor que (>) y menor que (<)
# .
numero = int(input("ingresa un numero: ")) # 1. Asignación de un número negativo.
if numero > 0: # 2. Primera condición: ¿es mayor a cero?
    print("El número es positivo.") # 3. Ejecuta si es mayor a 0.
elif numero < 0: # 4. Segunda condición: ¿es menor a cero?
    print("El número es negativo.") # 5. Ejecuta si es menor a 0.
else: # 6. Si no es mayor ni menor a cero.
    print("El número es cero.") # 7. Ejecuta si es exactamente 0.
```

```python
# 3. El condicional if-elif-else
# Permite evaluar múltiples condiciones en orden. En cuanto una se cumple, las demás se omiten
# .
# Ejemplo: Clasificación de un número (Positivo, Negativo o Cero) Utiliza los operadores relacionales mayor que (>) y menor que (<)
# .
numero = int(input("ingresa un numero: ")) # 1. Asignación de un número negativo.
if numero > 0: # 2. Primera condición: ¿es mayor a cero?
    print("El número es positivo.") # 3. Ejecuta si es mayor a 0.
elif numero < 0: # 4. Segunda condición: ¿es menor a cero?
    print("El número es negativo.") # 5. Ejecuta si es menor a 0.
else: # 6. Si no es mayor ni menor a cero.
    print("El número es cero.") # 7. Ejecuta si es exactamente 0.
```

```python
# 4. Uso de operadores lógicos (and, or, not)
# Estos operadores permiten combinar varias expresiones relacionales en una sola línea de decisión
# .
# Ejemplo: Verificar si un año es bisiesto Este ejemplo integra operadores aritméticos (%), relacionales (==, !=) y lógicos (and, or)
# .
año = int(input("3ingrese el año")) # 1. Se define el año a evaluar.
# 2. Un año es bisiesto si es divisible por 4 Y no por 100, O si es divisible por 400.
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print(f"{año} es bisiesto.") # 3. Si la lógica compleja es verdadera.
else:
    print(f"{año} no es bisiesto.") # 4. Si la lógica compleja es falsa.
```
